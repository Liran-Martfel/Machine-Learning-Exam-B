Q1 - Min-max normalise the column 
[18,42,66,90]. Show the value each of the four numbers is mapped to.

```python
x = np.array([18,42,66,90])
x_norm = (x-x.min()) / (x.max() - x.min())
print(x_norm)
```

Q2 - Find the standard deviation of the group of numbers 
[8,10,10,14,14,16]. Show the mean, every deviation from the mean, and the final result.

```python
data = np.array([8, 10, 10, 14, 14, 16])
data_mean = sum(data)/len(data)
deviations = data - data_mean
std = np.sqrt(((data - data_mean)**2).sum() / (len(data)))
print('std: ',std.round(4))
print('mean: ',data_mean)
print('deviations: ',deviations)
```

Give one advantage and one disadvantage of Min-Max normalisation, and one advantage and one disadvantage of Standardisation.

Advantage & DisAdvantage of Min-Max:
- it maps every value (no matter how far the value is from the others) into a range from 0-1.
- the Disadvantage is that the points can be very far from one another and it can impact the whole calculation (sensitive to outliers).

Advantage & DisAdvantage of standardisation:
- its less sensetive to outliers beacuse it uses the mean of every point.
- the Disadvantage is that there is no boundry to the range, and extreme outliers can still distort the mean and standard deviation.

# K-Nearest Neighbours

```python
min_support = 0
max_support = 10
min_usage = 0
max_usage = 100

x_norm_A_tickets = (2 - min_support) / (max_support - min_support)
x_norm_B_tickets = (9 - min_support) / (max_support - min_support)
x_norm_new_tickets = (6 - min_support) / (max_support - min_support)

x_norm_A_usage = (80 - min_usage) / (max_usage - min_usage)
x_norm_B_usage = (15 - min_usage) / (max_usage - min_usage)
x_norm_new_usage = (40 - min_usage) / (max_usage - min_usage)
print('Answer for A:')
print ('TICKETS:')
print('x_norm_A_tickets: ',x_norm_A_tickets)
print('x_norm_B_tickets: ',x_norm_B_tickets)
print('x_norm_new_tickets: ',x_norm_new_tickets)
print()

print ('USAGE:')
print('x_norm_A_usage: ',x_norm_A_usage)
print('x_norm_B_usage: ',x_norm_B_usage)
print('x_norm_new_usage: ',x_norm_new_usage)

```

```python
print('Answer to B:')

distance_new_to_a = np.sqrt(((x_norm_new_tickets - x_norm_A_tickets)**2) + (x_norm_new_usage - x_norm_A_usage)**2)
print('distance to A:',distance_new_to_a.round(4))

distance_new_to_b = np.sqrt(((x_norm_new_tickets - x_norm_B_tickets)**2) + (x_norm_new_usage - x_norm_B_usage)**2)
print('distance to B:',distance_new_to_b.round(4))

print()

print('Answer to C:')
if distance_new_to_a < distance_new_to_b:
    print('Closer to A: Predicted class: No (will not cancel)')
else:
    print('Closer to B 0: Predicted class: Yes (will cancel)')
```
