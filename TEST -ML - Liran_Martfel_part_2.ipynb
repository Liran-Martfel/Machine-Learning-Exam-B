# Decision Trees

```python
print('Answer to A:')
tickets = 14
escalated = 8
not_escalated = 6
Gini_parent = 1 - ((escalated/tickets)**2 + (not_escalated/tickets)**2)
print('Parent Gini:',np.round(Gini_parent,6))

print()

print('Answer to B:')
Child_A_tickets = 6
Child_B_tickets = 8
escalated_Child_A = 5
not_escalated_Child_A = 1
escalated_Child_B = 3
not_escalated_Child_B = 5

Gini_urgent = 1 - ((escalated_Child_A/Child_A_tickets)**2 + (not_escalated_Child_A/Child_A_tickets)**2)
print('Gini_urgent Gini:',np.round(Gini_urgent,6))
Gini_not_urgent = 1 - ((escalated_Child_B/Child_B_tickets)**2 + (not_escalated_Child_B/Child_B_tickets)**2)
print('Gini_not_urgent Gini:',np.round(Gini_not_urgent,6))

print()

print('Answer to C:')
Gini_split = ((Child_A_tickets/tickets) * Gini_urgent) + ((Child_B_tickets/tickets) * Gini_not_urgent)
Gain = Gini_parent - Gini_split
print('Weighted Gini: ',np.round(Gini_split,4))
print('Gain Gini: ',np.round(Gain,4))

print()

print('Answer to D:')
Rival_question = 0.08
if  Gain > Rival_question:
    print('The tree asks the "urgent" question')
else:
    print('The tree asks the Rival question')
```

```python
print('Answer to A:')
issue = np.array([1, 2, 4, 5, 7, 8])
premium = np.array([0, 1, 0, 1, 1, 1])
time = np.array([5, 7, 15, 17, 19, 21])

#because i don't have a y hat, and i have all the data, and i'm not trying to predict, i can assume that y hat = y bar
y_bar = time.sum() / len(time)
SSE = np.sum((time - y_bar)**2)
MSE = SSE / len(time)
print ('SSE: ',SSE)
print ('Y bar: ',y_bar)
print ('MSE: ',np.round(MSE,4))

print()

print('Answer to B:')
issue_over_3 = np.array([4, 5, 7, 8])
time_over_3 = np.array([15, 17, 19, 21])
issue_under_3 = np.array([1,2])
time_under_3 = np.array([5, 7])

## 3 or under
y_bar_under_3 = time_under_3.sum() / len(time_under_3)
SSE_under_3 = np.sum((time_under_3 - y_bar_under_3)**2)
MSE_under_3 = SSE_under_3 / len(time_under_3)
print ('SSE Under 3: ',SSE_under_3)
print ('Y bar Under 3: ',y_bar_under_3)
print ('MSE Under 3: ',np.round(MSE_under_3,4))

print()

## Higher than 3
y_bar_over_3 = time_over_3.sum() / len(time_over_3)
SSE_over_3 = np.sum((time_over_3 - y_bar_over_3)**2)
MSE_over_3 = SSE_over_3 / len(time_over_3)
print ('SSE Over 3: ',SSE_over_3)
print ('Y bar Over 3: ',y_bar_over_3)
print ('MSE Over 3: ',np.round(MSE_over_3,4))

print()

MSE_Split = ((len(issue_under_3) / len(issue)) * MSE_under_3) + ((len(issue_over_3) / len(issue)) * MSE_over_3)
Gain = MSE - MSE_Split
print(f'Weighted MSE: {np.round(MSE_Split,4)}\nGain: {np.round(Gain,4)}')

print()
```
